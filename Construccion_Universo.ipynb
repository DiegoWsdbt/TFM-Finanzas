```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import yfinance as yf
```

```python
START_DATE = "2010-01-01"
END_DATE = "2024-12-31"
N = 50
WINDOW = 60

sp500_hist = pd.read_csv("sp500_historical.csv")
sp500_2009 = sp500_hist[
    sp500_hist["date"] == "2009-12-31"
]
sp500_2009["ticker"]
```

```python
universe_50 = (
    sp500_2009
    .sort_values("weight", ascending=False)
    .head(54)
    .copy()
)
display(universe_50[["rank", "ticker", "name", "sector", "weight"]].head())
print("Número de acciones:", len(universe_50))
print("Número de sectores:", universe_50["sector"].nunique())
display(universe_50["sector"].value_counts())
```

```python
ticker_mapping = {"AEC1": "AXP", "0R01": "C", "GEC": "GE", "SLBA": "SLB", "UTX": "RTX", "VISA": "V"}

universe_50["ticker_yf"] = (universe_50["ticker"].str.replace("*", "", regex=False).str.strip().replace(ticker_mapping))
universe_50 = universe_50[(universe_50["ticker_yf"] != "KFT") & (universe_50["ticker_yf"] != "MON") & (universe_50["ticker_yf"] != "INCO") & (universe_50["ticker_yf"] != "QCI")].copy()
display(universe_50.head())
```

```python
special_tickers = universe_50[
    ~universe_50["ticker"].str.match(r"^[A-Z]+$")
]

display(
    special_tickers[
        ["rank", "ticker", "ticker_yf", "name"]
    ]
)
```

```python
universe_50.to_csv("universe_50_2009.csv", index=False)
```
